# Exploratory Data Analysis

```python
# Library
import pandas as pd
import plotly.express as px
```

## Loading the dataset

```python

data = pd.read_csv("data/tracks.csv")
data = data.iloc[2:]
data.head()
```

## Exploring distributions
Before visualising data such as favourites and comments, we'd like to know if there's enough data

```python
fig1 = px.histogram(data, x="album.5", title = "Distribution of album favourites")
fig1.show()

fig2 = px.histogram(data, x="album", title = "Distribution of album comments")
fig2.show()
```
