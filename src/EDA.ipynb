## Loading the dataset

```python

data = pd.read_csv("../data/tracks.csv", index_col=0)

# Merge the first row with the header, to make the dataset easier to work with
first_row = data.iloc[0].astype(str)
headers = data.columns.astype(str)
data.columns = [f"{h}_{r}" if h != r else h for h, r in zip(headers, first_row)]
data = data[1:].reset_index(drop=True)
data = data.iloc[1:]
data.head()
```

## Checking completeness
We need to check the completeness of the data to see how "bad" the damage is

```python
# The following function 'missing_values_table' was sourced from a GitHub Gist.
# Author: clarissaache
# Gist URL: https://gist.github.com/clarissaache/completeness.py (derived from the image)
# Date Accessed: May 10, 2026
def missing_values_table(df):
    ''' get the total count and % of NaN in a cute little table'''
    mis_val = df.isnull().sum()
    mis_val_percent = 100 * df.isnull().sum() / len(df)
    mis_val_table = pd.concat([mis_val, mis_val_percent], axis=1)
    mis_val_table_ren_columns = mis_val_table.rename(
    columns = {0 : 'Missing Values', 1 : '% of Total Values'})
    mis_val_table_ren_columns = mis_val_table_ren_columns[
        mis_val_table_ren_columns.iloc[:,1] != 0].sort_values(
    '% of Total Values', ascending=False).round(1)
    print ("Your selected dataframe has " + str(df.shape[1]) + " columns.\n"      
        "There are " + str(mis_val_table_ren_columns.shape[0]) +
            " columns that have missing values.")
    return mis_val_table_ren_columns
```

```python
missing_values_table(data)
```

This only shows the percentage of missin values across all rows, but not per individual song, album or artist

```python
# Information we'd like to have

def count_empty(df, cols: List, id: str):
    """
    Function takes dataset, list of target columns and the column name of the ID
    :param df: dataset of type dataframe
    :param cols: list of columns of list type
    :param id: name of the column where the key identifier is of string type
    :returns: a summary of how the percentages of empty values per unique ID
    """
    if id == "index":
        key_id = df.groupby(df.index)[cols].first()
    else:
        key_id = df.groupby(id)[cols].first()
    null_counts = key_id.isna().sum()
    total_key_id = len(key_id)
    null_percentages = (null_counts / total_key_id) * 100
    summary = pd.DataFrame({
    'Empty (Count)': null_counts,
    'Empty (%)': null_percentages.round(2)
    })

    print(summary.sort_values('Empty (%)', ascending=False))


```

```python
album_cols = [
    'album.1_date_created', 'album.2_date_released', 
    'album.3_engineer', 'album.6_information', 
    'album.7_listens', 'album.8_producer', 'album.9_tags', 
    'album.10_title', 'album.11_tracks', 'album.12_type'
]

artist_cols = [
    "artist.1_active_year_end", "artist.2_associated_labels",
    "artist.3_bio", "artist.5_date_created",
    "artist.6_favorites", "artist.8_latitude",
    "artist.9_location", "artist.10_longitude",
    "artist.11_members", "artist.12_name"
    
]

track_cols = [
    "track_bit_rate", "track.2_composer",
    "track.3_date_created", "track.4_date_recorded",
    "track.5_duration", "track.7_genre_top",
    "track.8_genres", "track.9_genres_all",
    "track.10_information",	"track.11_interest",
    "track.12_language_code", "track.13_license",
    "track.14_listens", "track.15_lyricist",
    "track.16_number", "track.17_publisher",
    "track.18_tags", "track.19_title"
    
]


count_empty(df = data, cols = album_cols, id = "album.5_id")
count_empty(df = data, cols = artist_cols, id = "artist.7_id")
count_empty(df = data, cols = track_cols, id = "index")

```
